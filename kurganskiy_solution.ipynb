Из-за того, что функции numpy по умолчанию используют фиксированные типы данных, для больших чисел мы не сможем получить правильный ответ

**Японский кроссворд**

```python
def crossword(rows, cols, fill_prob=0.6, seed=None):
    rng = np.random.default_rng(seed)
    grid = (rng.random((rows, cols)) < fill_prob).astype(int)

    def blocks_lengths(line):
        padded = np.concatenate([[0], line, [0]])
        diffs = np.diff(padded)
        starts = np.where(diffs == 1)[0]
        ends   = np.where(diffs == -1)[0]
        return (ends - starts).tolist()

    row_keys = [blocks_lengths(grid[i]) for i in range(rows)]
    col_keys = [blocks_lengths(grid[:, j]) for j in range(cols)]

    return grid, row_keys, col_keys
```

```python
grid, rk, ck = crossword(5, 5, seed=24)
print("Матрица:\n", grid)
print("Ключи строк:   ", rk)
print("Ключи столбцов:", ck)
```

**Расширенный алгоритм Евклида**

```python
def ex_euclid(a, b):
    # Чтобы int64 не переполнился на больших числах, используем dtype=object
    coef = np.array([[1, 0], [0, 1]], dtype=object)  
    vals = np.array([a, b], dtype=object)      

    while vals[1] != 0:
        q = vals[0] // vals[1]                 
        step = np.array([[0, 1], [1, -q]], dtype=object)   
        coef = step @ coef                     
        vals = step @ vals                     

    gcd = int(vals[0])
    x, y = coef[0]   
    print(f"НОД: {g}, множитель при а: {x}, множитель при b: {y}. Тест: {a} * {x} + {b} * {y} = {a * x + b * y}")
    return gcd, int(x), int(y)
```

**Усложнённая игра в жизнь**

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
%matplotlib widget 

M = 101                
MAX_AGE = 5            # сколько поколений живёт клетка
MUTANT_PROB = 0.0005   # вероятность мутанта на клетку за шаг
DENSITY_HIGH = 0.6     # порог высокой плотности
DENSITY_LOW = 0.15     # порог низкой плотности

rng = np.random.default_rng(42)
X = rng.choice([1, 0], M * M, p=[0.2, 0.8]).reshape(M, M).astype(np.int8)
ages = np.zeros((M, M), dtype=np.int16)

def count_neighbors_8(grid):
    p = np.pad(grid, 1)   
    N = (p[:-2, :-2] + p[1:-1, :-2] + p[2:, :-2]      
       + p[:-2, 1:-1] +                   p[2:, 1:-1] 
       + p[:-2, 2:]  + p[1:-1, 2:]  + p[2:, 2:])      
    return N

def count_density_5x5(grid):
    p = np.pad(grid, 2)   # рамка нулеё шириной 2
    total = np.zeros_like(grid, dtype=np.int16)
    for di in range(-2, 3):
        for dj in range(-2, 3):
            if di == 0 and dj == 0:
                continue
            total += p[2 + di : 2 + di + M, 2 + dj : 2 + dj + M]
    return total / 24.0   


def at_frame(frame, image, X, ages):
    n8  = count_neighbors_8(X)
    den = count_density_5x5(X)

    # Классические правила 
    survive_classic = X & ((n8 == 2) | (n8 == 3))
    born_classic    = (~X.astype(bool)) & (n8 == 3)
    X_new = (survive_classic | born_classic).astype(np.int8)

    # Иммунитет по плотности
    high = den > DENSITY_HIGH
    low  = den < DENSITY_LOW
    mid  = ~(high | low)

    # Иммунитет при средней плотности
    immune_survive = mid & X.astype(bool) & (n8 >= 1) & (n8 <= 4)
    # Низкая плотность
    low_survive    = low & X.astype(bool) & (n8 == 3)
    X_new = (X_new | immune_survive | low_survive).astype(np.int8)
    # Высокая плотность
    X_new[high] = 0

    # Время жизни
    ages = np.where(X_new == 1, ages + 1, 0).astype(np.int16)
    too_old = ages > MAX_AGE
    X_new[too_old] = 0
    ages[too_old] = 0

    # Мутанты
    mutants = rng.random((M, M)) < MUTANT_PROB
    X_new[mutants] ^= 1          # пускай мутант меняет своё состояние

    X[:] = X_new
    ages[:] = ages

    image.set_data(X)
    return image,

fig, ax = plt.subplots(figsize=(6, 6))
image = ax.imshow(X, interpolation='nearest', cmap='viridis')
ax.set_title("Расширенная игра «Жизнь»")

ani = animation.FuncAnimation(
    fig, at_frame,
    fargs=(image, X, ages),
    frames=300, interval=100, blit=True
)
plt.show()
```

Модель линейной регрессии в самом простом виде предполагает следующее.

Пусть заданы два набора числовых данных $X=\{x_1,x_2,...x_k\}$ и $Y=\{y_1,y_2,...y_k\}$. Необходимо по известным данным отыскать такую функцию вида $f(x) = a*x +b $ максимально описывающую зависимость представителей набора $Y$ от представителей набора $X$.

Задача обучения в данном случае состоит в нахождении коэффициентов $a$ и $b$ таких, чтобы минимизировать значение квадратичной ошибки $SSE(a,b) = \displaystyle\sum_{i=1}^{k}(y_i - f(x_i))^2$ на исходных данных. 

Разрешив полученную задачу оптимизации можно получить формулы для коэффициентов $a$ и $b$.

$$a = \frac{\displaystyle\sum_{i=1}^{k}(x_k - \overline{x})(y_k - \overline{y})}{\displaystyle\sum_{i=1}^{k}(x_k - \overline{x})^2}$$

$$b = \overline{y}- a \overline{x} $$

Полученные формулы можно представить в **матричном виде**, если записать исходные данные в вектор столбцы или вектор строки.

**Пункт 1**

Обычные формулы:

$$a = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}, \qquad b = \bar{y} - a\bar{x}$$

Обозначим:
- $X_c = X - \bar{X}$ 
- $Y_c = Y - \bar{Y}$ 

Тогда:

$$a = \frac{X_c^T Y_c}{X_c^T X_c}, \qquad b = \bar{y} - a\bar{x}$$

**Пункт 2:**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import numpy as np

class MyLinearRegression:
    def fit(self, X, y):
        X = np.asarray(X, dtype=float).ravel()
        y = np.asarray(y, dtype=float).ravel()

        Xc = X - X.mean()          
        yc = y - y.mean()         

        # a = X_c^T Y_c / X_c^T X_c
        a = np.dot(Xc, yc) / np.dot(Xc, Xc)
        # b = ȳ - a x̄
        b = y.mean() - a * X.mean()

        self.a_, self.b_ = a, b
        return a, b

    def predict(self, X):
        """Предсказание: ŷ = a*x + b."""
        return self.a_ * np.asarray(X) + self.b_

    def score(self, X, y_true):
        """R² — коэффициент детерминации."""
        y_true = np.asarray(y_true, dtype=float)
        y_pred = self.predict(X)
        ss_res = np.sum((y_true - y_pred) ** 2)
        ss_tot = np.sum((y_true - y.mean()) ** 2)
        return 1 - ss_res / ss_tot
```
